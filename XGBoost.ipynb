```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from xgboost import XGBRegressor
df = pd.read_csv(
    "preprocessed_data.csv"
)

print("Dataset Loaded Successfully!")
print("Total Vehicles:", len(df))

X = df[
    [
        "battery_capacity_kwh",
        "state_of_charge",
        "battery_health",
        "energy_consumption_kwh_per_100km",
        "vehicle_age",
        "average_speed",
        "vehicle_weight_kg"
    ]
]

y = df["estimated_range_km"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("\nXGBoost Model Trained Successfully!")
y_pred = model.predict(X_test)

y_pred = np.round(y_pred, 2)
r2 = r2_score(y_test, y_pred)

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

accuracy = r2 * 100

print("\n======================================")
print("        XGBOOST REGRESSION")
print("======================================")

print(
    f"\nR² Score: {r2:.4f}"
)

print(
    f"Accuracy / R² Percentage: {accuracy:.2f}%"
)

print(
    f"Mean Absolute Error: {mae:.2f} km"
)

print(
    f"Root Mean Squared Error: {rmse:.2f} km"
)

result = pd.DataFrame({
    "Actual Range (km)": y_test.values,
    "Predicted Range (km)": y_pred
})

print("\nActual vs Predicted Range")
print("--------------------------------")

print(result.head(20))
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

print("\nFeature Importance")
print("--------------------------------")

print(importance)
```
